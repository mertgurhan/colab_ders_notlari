#KNN_sklearn



```python
import pandas as pd
```

```python
data = pd.read_csv("iris.csv", names=["s1","sw","pl","pw","class"])

data.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

K-En Yakın Komşu (KNN) sınıflandırıcısıdır. Yeni bir örneğin etiketini, eğitim verisindeki en yakın k komşusuna bakarak tahmin eder (ör. k=5).

```python
Y=data.pop("class")
X=data
```

```python
train_x, test_x, train_y, test_y =train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
mms=StandardScaler()
mms.fit(train_x)
train_x=mms.transform(train_x)
test_x=mms.transform(test_x)

#eğitim verisini dönüştürüyor
```

```python
knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(train_x,train_y)
```

```python
knn_model.predict([[6.3,2.5,5.0,1.9]])
```

```python
len(test_x)
```

```python
predicted_y=knn_model.predict(test_x)
predicted_y
```

```python
for i in predicted_y:
  pass
true_counter=0
index_counter=0
for prediction,real in zip(predicted_y,test_y):
  if prediction==real:
    true_counter+=1
  else:
    print(prediction,real,index_counter)
    index_counter+=1
```

```python
total_test_samples=len(test_y)
oran=round((true_counter/total_test_samples)*100,2)
print("Oran:%",oran)
```

```python
data = pd.read_csv("iris.csv", names=["sl","sw","pl","pw","class"])
data.head()
```

```python
sample_1 = [4.8,3.0,1.4,0.1] # setosa
sample_2 = [6.4,2.9,4.3,1.3] # versicolor
sample_3 = [6.3,2.5,5.0,1.9] # virgina

K = 3
```

```python
def calculate_distance(ins1,ins2):
  sum=0
  for index in range(len(ins1)):
    sum+=(ins1[index]-ins2[index])**2
  sqr=sum**0.5
  return sqr
```

```python
#test distance function
sample_a = [1,1,2,3,5]
sample_b = [2,2,2,3,4]
calculate_distance(sample_a,sample_b)
```

```python
print("Sınıflandırma : ",sample_2 )
distance_list=[]
for index,row in data.iterrows():
  data_sample=row.values[:-1]
  true_class=row.values[-1]
  distance=calculate_distance(data_sample,sample_2)
  print(data_sample,true_class,distance)
  distance_list.append([distance,true_class,distance])
```
