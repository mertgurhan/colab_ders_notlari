```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns


# Excel dosyalarından öğrenci verilerini oku
def read_student_data(folder_path):
    all_student_data = []
    files = os.listdir(folder_path)
    for file in files:
        if file.endswith('.xlsx'):
            file_path = os.path.join(folder_path, file)
            # Dosyayı oku
            data = pd.read_excel(file_path)
            # Dosyadan gelen verilere öğrenci adını ekle
            data['Öğrenci'] = os.path.splitext(file)[0]
            # Ders isimlerini 3. sütundan al
            data['ders'] = data.iloc[:, 2]
            all_student_data.append(data)
    # Tüm öğrenci verilerini birleştir
    student_data = pd.concat(all_student_data, ignore_index=True)
    return student_data


# Ortalama ve Genel Ortalama sütunlarını hesapla
def calculate_averages(student_data):
    # Her ders için ayrı ayrı genel ortalama hesapla
    for course_name, group in student_data.groupby('ders'):
        total_average = 0
        total_students = 0

        # Her öğrencinin not ortalamasını tek tek hesapla ve topla
        for student_name, student_group in group.groupby('Öğrenci'):
            student_average = student_group['vize'] * 0.4 + student_group['final'] * 0.6
            total_average += student_average.mean()
            total_students += 1

        # Tüm öğrencilerin not ortalamalarının ortalamasını alarak genel ortalama hesapla
        overall_average = total_average / total_students
        student_data.loc[student_data['ders'] == course_name, 'Ortalama'] = group['vize'] * 0.4 + group['final'] * 0.6
        student_data.loc[student_data['ders'] == course_name, 'Genel_Ortalama'] = overall_average


# Öğrenci verilerini görselleştirme
def visualize_student_data(student_data, student_name=None, class_name=None, course_name=None):
    # Filtreleme yap
    filtered_data = student_data.copy()
    if student_name:
        filtered_data = filtered_data[filtered_data['Öğrenci'] == student_name]
    if class_name:
        filtered_data = filtered_data[filtered_data['sınıf'] == class_name]
    if course_name:
        filtered_data = filtered_data[filtered_data['ders'] == course_name]

    # Her ders için genel ortalama
    overall_course_averages = filtered_data.groupby('ders')['Ortalama'].mean()

    # Grafikleri görselleştir
    visualize_line_plot(filtered_data, overall_course_averages)
    visualize_scatter_plot(filtered_data, overall_course_averages)
    visualize_histogram(filtered_data, overall_course_averages)
    visualize_bar_plot(filtered_data, overall_course_averages)
    visualize_box_plot(filtered_data, overall_course_averages)
    visualize_heatmap(filtered_data, overall_course_averages)


# Çizgi grafiği görselleştirme
def visualize_line_plot(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    sns.lineplot(data=filtered_data, x='ders', y='Ortalama', hue='Öğrenci', marker='o')
    # Genel ortalama ekle
    for course, overall_average in overall_course_averages.items():
        plt.axhline(y=overall_average, color='gray', linestyle='--', label=f'{course} Genel Ortalama')
    plt.title('Ders Ortalama Karşılaştırması')
    plt.xlabel('Ders')
    plt.ylabel('Ortalama')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.legend(title='Öğrenci')
    plt.show()


# Saçılım grafiği görselleştirme
def visualize_scatter_plot(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=filtered_data, x='Ortalama', y='Genel_Ortalama', hue='ders')
    # Genel ortalama ekle
    for course, overall_average in overall_course_averages.items():
        plt.axhline(y=overall_average, color='gray', linestyle='--', label=f'{course} Genel Ortalama')
    plt.title('Öğrenci Ortalamaları ve Genel Ortalama Arasındaki İlişki')
    plt.xlabel('Ortalama')
    plt.ylabel('Genel Ortalama')
    plt.tight_layout()
    plt.legend(title='Ders')
    plt.show()


# Diğer görselleştirme fonksiyonlarını da benzer şekilde güncelleyebilirsiniz.

def main():
    # Öğrenci verilerini oku
    folder_path = r"C:\Users\Sude\Desktop\VBG\08_hafta_1\08.hafta"
    student_data = read_student_data(folder_path)

    # Ortalama ve Genel Ortalama sütunlarını hesapla
    calculate_averages(student_data)

    # Kullanıcıdan isim al
    student_name = input("Lütfen öğrenci adını girin: ")

    # Öğrenci verilerini ekrana yazdır
    filtered_student_data = student_data[student_data['Öğrenci'] == student_name]
    if not filtered_student_data.empty:
        print(filtered_student_data[
                  ['sno', 'sınıf', 'ders', 'vize', 'final', 'Ortalama', 'Genel_Ortalama', 'Öğrenci']].head())
        # Öğrenci verilerini görselleştir
        visualize_student_data(student_data, student_name=student_name)
    else:
        print("Girdiğiniz öğrenci adına ait veri bulunamadı.")

# Histogram grafiği görselleştirme
def visualize_histogram(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    sns.histplot(data=filtered_data, x='Ortalama', bins=10, kde=True, hue='ders')
    # Genel ortalama ekle
    for course, overall_average in overall_course_averages.items():
        plt.axvline(x=overall_average, color='gray', linestyle='--', label=f'{course} Genel Ortalama')
    plt.title('Öğrenci Ortalamalarının Dağılımı')
    plt.xlabel('Ortalama')
    plt.ylabel('Frekans')
    plt.tight_layout()
    plt.legend(title='Ders')
    plt.show()


# Sütun/Çubuk grafiği görselleştirme
def visualize_bar_plot(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    sns.barplot(data=filtered_data, x='ders', y='Ortalama', hue='Öğrenci')
    # Genel ortalama ekle
    for course, overall_average in overall_course_averages.items():
        plt.axhline(y=overall_average, color='gray', linestyle='--', label=f'{course} Genel Ortalama')
    plt.title('Öğrenci Ortalamaları')
    plt.xlabel('Ders')
    plt.ylabel('Ortalama')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.legend(title='Öğrenci')
    plt.show()


# Kutu grafiği (BoxPlot) görselleştirme
def visualize_box_plot(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=filtered_data, x='ders', y='Ortalama', hue='Öğrenci')
    # Genel ortalama ekle
    for course, overall_average in overall_course_averages.items():
        plt.axhline(y=overall_average, color='gray', linestyle='--', label=f'{course} Genel Ortalama')
    plt.title('Ders Ortalamalarının Dağılımı')
    plt.xlabel('Ders')
    plt.ylabel('Ortalama')
    plt.tight_layout()
    plt.legend(title='Öğrenci')
    plt.show()


# Isı haritası grafiği görselleştirme
def visualize_heatmap(filtered_data, overall_course_averages):
    plt.figure(figsize=(10, 6))
    pivot_table = filtered_data.pivot_table(index='Öğrenci', columns='ders', values='Ortalama')
    sns.heatmap(pivot_table, cmap='viridis', annot=True, fmt=".1f")
    plt.title('Ders Ortalamalarının Isı Haritası')
    plt.xlabel('Ders')
    plt.ylabel('Öğrenci')
    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    main()
```
