```python
import random
import math

# Örnek maliyet matrisi (işçi-görev maliyetleri)
cost_matrix = [
    [9, 2, 7, 8],
    [6, 4, 3, 7],
    [5, 8, 1, 8],
    [7, 6, 9, 4]
]
def calculate_cost(solution, cost_matrix):
    return sum(cost_matrix[i][solution[i]] for i in range(len(solution)))
def random_solution(size):
    solution = list(range(size))
    random.shuffle(solution)
    return solution

def neighbor_solution(solution):
    neighbor = solution[:]
    i, j = random.sample(range(len(solution)), 2)
    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
    return neighbor

def simulated_annealing(cost_matrix, initial_temp, cooling_rate, max_iterations):
    size = len(cost_matrix)
    current_solution = random_solution(size)
    current_cost = calculate_cost(current_solution, cost_matrix)
    best_solution = current_solution[:]
    best_cost = current_cost
    temperature = initial_temp
    for _ in range(max_iterations):
        if temperature <= 0:
            break
        new_solution = neighbor_solution(current_solution)
        new_cost = calculate_cost(new_solution, cost_matrix)

        if new_cost < current_cost or random.random() < math.exp((current_cost - new_cost) / temperature):
            current_solution, current_cost = new_solution, new_cost

        if current_cost < best_cost:
            best_solution, best_cost = current_solution[:], current_cost

        temperature *= cooling_rate

    return best_solution, best_cost
initial_temp = 1000
cooling_rate = 0.95
max_iterations = 1000
best_solution, best_cost = simulated_annealing(cost_matrix, initial_temp, cooling_rate, max_iterations)
print("En iyi çözüm:", best_solution)
print("Toplam maliyet:", best_cost)
```

```python

import matplotlib.pyplot as plt
import numpy as np
# (amaç fonksiyonu)
def rastrigin(position):
    A = 10
    return A * len(position) + sum([(x ** 2 - A * np.cos(2 * np.pi * x)) for x in position])
def generate_population(size, dimensions, bounds):
    return np.random.uniform(bounds[0], bounds[1], (size, dimensions))

# Black Hole Algoritması
def black_hole_algorithm(objective_function, dimensions, bounds, population_size=30, max_iterations=100):
    # Başlangıç popülasyonu
    population = generate_population(population_size, dimensions, bounds)
    fitness = np.array([objective_function(ind) for ind in population])
    best_index = np.argmin(fitness)
    black_hole = population[best_index]
    black_hole_fitness = fitness[best_index]

    # İterasyon döngüsü
    for iteration in range(max_iterations):
        # Yıldızları kara deliğe doğru çek
        for i in range(population_size):
            if i == best_index:
                continue  # Kara delik kendine çekim uygulamaz
            r = np.random.random(dimensions)  # Rastgele çekim faktörü
            population[i] += r * (black_hole - population[i])
        population = np.clip(population, bounds[0], bounds[1])
        fitness = np.array([objective_function(ind) for ind in population])

        new_best_index = np.argmin(fitness)
        new_black_hole_fitness = fitness[new_best_index]

        if new_black_hole_fitness < black_hole_fitness:
            black_hole = population[new_best_index]
            black_hole_fitness = new_black_hole_fitness
            best_index = new_best_index

        # Yıldızları olay ufkunun dışına çıkar
        event_horizon = np.mean(fitness)
        for i in range(population_size):
            if fitness[i] > event_horizon:
                population[i] = np.random.uniform(bounds[0], bounds[1], dimensions)
                fitness[i] = objective_function(population[i])
        print(f"Iteration {iteration + 1}/{max_iterations}: Best Fitness = {black_hole_fitness}")
    return black_hole, black_hole_fitness
dimensions = 2
bounds = [-5.12, 5.12]  # Değişken sınırları (Rastrigin fonksiyonu için)
population_size = 30
max_iterations = 100
best_solution, best_fitness = black_hole_algorithm(rastrigin, dimensions, bounds, population_size, max_iterations)
print("\nOptimal Çözüm:", best_solution)
print("Optimal Fitness (En Küçük Değer):", best_fitness)
```
