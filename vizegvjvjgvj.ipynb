```python
import pandas as pd
import matplotlib.pyplot as plt
import sys

ogrenci = input("bilgilerini istediğiniz öğrencinin isim-soyisim girin")

# dosya ismini bulmak için isimi düzenleme    #
isim, soyisim = ogrenci.split()
ogrenci1 = isim + soyisim
ogrenci1 = ogrenci1.lower() # HATA YOK !

#ogrenci varmı yok mu onu arıyor

if ogrenci1 == "alikaya":
    veri = pd.read_excel("alikaya.xlsx")
elif ogrenci1 == "aynurtok":
    veri = pd.read_excel("aynurtok.xlsx")
elif ogrenci1 == "ayseak":
    veri = pd.read_excel("ayseak.xlsx")
elif ogrenci1 == "burakdeniz":
    veri = pd.read_excel("burakdeniz.xlsx")
elif ogrenci1 == "haliltorun":
    veri = pd.read_excel("haliltorun.xlsx")
elif ogrenci1 == "kadircelik":
    veri = pd.read_excel("kadircelik.xlsx")
elif ogrenci1 == "vedatbilir":
    veri = pd.read_excel("vedatbilir.xlsx")
elif ogrenci1 == "veyselkara":
    veri = pd.read_excel("veyselkara.xlsx")
elif ogrenci1 == "zelihauyur":
    veri = pd.read_excel("zelihauyur.xlsx")
else:
    print("ÖGRENCİ BULUNAMADI !!") #HATA YOK !!
    sys.exit()


secim = input("ogrenciye ait hangi veriyi istiyorsunuz (vize/final/ort)")

if secim == "vize":
    plt.plot(veri['ders'], veri['vize'], marker='.', linestyle='-')
    plt.title('Öğrenci Notları')
    plt.xlabel('DERSLER')
    plt.ylabel('VİZE NOTLARI')
    plt.xticks(rotation=90)
    plt.yticks(veri['vize'])
    plt.show()
elif secim == "final":
    plt.plot(veri['ders'], veri['final'], marker='.', linestyle='-')
    plt.title('Öğrenci Notları')
    plt.xlabel('DERSLER')
    plt.ylabel('FİNAL NOTLARI')
    plt.xticks(rotation=90)
    plt.yticks(veri['final'])
    plt.show()

elif secim == "ort":
    ort = (veri['vize']*0.4 + veri['final']*0.6)
    plt.plot(veri['ders'], ort, marker='.', linestyle='-')
    plt.title('Ders Ortalamaları')
    plt.xlabel('DERSLER')
    plt.ylabel('ORTALAMA')
    plt.xticks(rotation=90)
    plt.yticks(ort)
    plt.show()
else:
    print("lutfen istediginiz veriyi dogru girin")
```
