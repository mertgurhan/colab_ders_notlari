```python
#kirpma
import tkinter as tk
from PIL import Image, ImageTk
import os

output_folder = r"C:\Users\rumus\OneDrive\Masaüstü\folder3"
image_path = r"C:\Users\rumus\OneDrive\Masaüstü\imge2\r1.png"

# Sol üst dikdörtgenin koordinatları: (x, y)
left_top = (40, 26)
right_top = (86, 26)
left_bottom = (40, 49)

left = left_top[0]  #x1
upper = left_top[1]  #y1
right = right_top[0]  #x2
lower = left_bottom[1]  #y2
width = right_top[0] - left_top[0]  #genişlik
height = left_bottom[1] - left_top[1]  #yükseklik

matris= '11x5'
matrix_y = int(matris.split("x")[0])
matrix_x = int(matris.split("x")[1])

a2_lt = (86, 26)
other_next_lt = (40, 49)
order = 2
line_pix=1
x_in = 0  # Yatay düzlemde içeri girilen piksel miktarı (soldan x, sağdan x)
y_in = 0  # Dikey düzlemde içeri girilen piksel miktarı (yukarıdan y, aşağıdan y)
if image_path.endswith(".png"):
    image = Image.open(image_path)
    image_basename = os.path.basename(image_path)    # Resim adını ve uzantısını al
    image_name, image_ext = os.path.splitext(image_basename)
    image_name = image_name.replace("r", "resim")    # "r" harfini "resim" ile değiştir
    id=0.1
    if order==2 :
        hop = ((a2_lt[0] - right),(other_next_lt[1] - lower))
        hop_x = hop[0]
        hop_y = hop[1]
        for i in range(matrix_y):
            lower = upper + height
            for j in range(matrix_x):
                right = left + width
                region = image.crop((left + x_in, upper + y_in, (right + line_pix) - x_in, (lower + line_pix) - y_in))
                filename = os.path.join(output_folder, f"{image_name}_{id}{image_ext}")
                region.save(filename)

                print(f"sol:{left} sağ:{right} üst:{upper} alt:{lower}")
                print(f"{filename} kaydedildi.")
                left = right + hop_x
                id+=0.1
            upper = lower + hop_y
            left = left_top[0]

    if order==1 :
        hop = ((other_next_lt[0] - right),(a2_lt[1] - lower))
        hop_x = hop[0]
        hop_y = hop[1]
        for j in range(matrix_x):
            right = left + width
            for i in range(matrix_y):
                lower = upper + height
                region = image.crop((left + x_in, upper + y_in, (right + line_pix) - x_in, (lower + line_pix) - y_in)) #çizgiler gözüküyor
                filename = os.path.join(output_folder, f"{image_name}_{id}{image_ext}")
                region.save(filename)
                print(f"sol:{left} sağ:{right} üst:{upper} alt:{lower}")
                print(f"{filename} kaydedildi.")
                upper = lower + hop_y
                id+=0.1
            left = right + hop_x
            upper = left_top[1]
```

```python
birlestirme
import tkinter as tk
from PIL import Image, ImageTk
import os

folder_path = r"C:\Users\rumus\OneDrive\Masaüstü\folder3"
output_folder = r"C:\Users\rumus\OneDrive\Masaüstü\folder3"
photos=[os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(('.png'))]

# Sol üst dikdörtgenin koordinatları: (x, y)
left_top = (40, 26)
left = left_top[0]
upper = left_top[1]

matris= '11x5'
matrix_y = int(matris.split("x")[0])
matrix_x = int(matris.split("x")[1])

line_pix = 1  #alt ve sağ çizgi göeünsün/görünmesin
order = 2   #2 yatay 1 dikey
hop = (0,0)
hop_x= hop[0]
hop_y= hop[1]

filename = os.listdir(folder_path)[0]
file_path = os.path.join(folder_path, filename)
image = Image.open(file_path)
width, length = image.size

width-=line_pix  #doğru ölçütlerde yerleştirilmesi için gen ve yük'ten -1
length-=line_pix

new_w = (width*matrix_x) + ((matrix_x-1)*hop_x)
new_l = (length*matrix_y) + ((matrix_y-1)*hop_y)

image_basename = os.path.basename(file_path)    # Resim adını ve uzantısını al
image_name, image_ext = os.path.splitext(image_basename)
image_name = image_name.split("_")[0]
image_name = image_name.replace("resim", "r1")
new_img = Image.new("RGB", (new_w + (2 * left), new_l + (2 * upper)), "white")
id=0   #her bir foto için id
if order == 1:
    for i in range(matrix_x):
        right= left + width
        for j in range(matrix_y):
            lower= upper + length
            photo=Image.open(photos[id])
            new_img.paste(photo, (left, upper))
            upper = lower + hop_y
            id+=1
        left= right + hop_x
        upper = left_top[1]
elif order==2:
    for j in range(matrix_y):
        lower= upper + length
        for i in range(matrix_x):
            right= left + width
            photo=Image.open(photos[id])
            new_img.paste(photo, (left, upper))
            left= right + hop_x
            id+=1
        upper = lower + hop_y
        left = left_top[0]

output_path = os.path.join(output_folder, f"{image_name}{image_ext}")
new_img.save(output_path)
print(f"{image_name}{image_ext}, {output_folder} dosyasına kaydedildi")
```

```python
#olusturma
import tkinter as tk
from PIL import Image, ImageTk, ImageDraw, ImageFont
import os
folder = r"C:\Users\rumus\OneDrive\Masaüstü\folder3"
coor=(40, 26)
left= coor[0]
up = coor[1]
width,length=(46, 23)
gap_x, gap_y = (0,23)
line = 1
line_pix_x, line_pix_y=(0,0)
matrix_x, matrix_y = (5,6)
order = 2
no=1
letter="C"
if letter=="C":
    ext=3
elif letter =="A":
    ext=1
elif letter =="B":
    ext=2
elif letter=="D":
    ext=4
image = Image.new("RGB",((matrix_x*width)+((matrix_x-1)*gap_x)+(2*coor[0]),(matrix_y*length)+((matrix_y-1)*gap_y)+(2*coor[1])), "white")
#image.show()
draw= ImageDraw.Draw(image)
font = ImageFont.load_default()
if order ==1:
    for i in range(matrix_x):
        right= left+width
        for j in range(matrix_y):
            down=up+length
            draw.rectangle([left,up,right,down], outline="black", width=line)
            text = f"{letter}{no}"
            text_width = draw.textlength(text)
            text_heigth= font.getmask(text).getbbox()[3]
            text_x = left + (width - text_width) / 2
            text_y = up + (length - text_heigth) / 2
            draw.text((text_x, text_y), text, fill="black", font=font)
            no+=1
            up= down+gap_y
        left=right+gap_x
        up=coor[1]

elif order ==2:
    for j in range(matrix_y):
        down=up+length
        for i in range(matrix_x):
            right=left+width
            draw.rectangle([left,up,right,down], outline="black", width=line)
            text = f"{letter}{no}"
            text_width = draw.textlength(text)
            text_heigth= font.getmask(text).getbbox()[3]
            text_x = left + (width - text_width) / 2
            text_y = up + (length - text_heigth) / 2
            draw.text((text_x, text_y), text, fill="black", font=font)
            no+=1
            left=right+ gap_x
        up=down+gap_y
        left= coor[0]
file_path= os.path.join(folder, f"r2{ext}.png")
image.save(file_path)
print(f"{os.path.basename(file_path)} kaydedildi")
image.show()
```
