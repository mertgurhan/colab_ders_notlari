```python
import pytesseract
import cv2
import fitz  # PyMuPDF
import pytesseract
from pdf2image import convert_from_path
from PIL import Image
import os

def pdf_to_text_with_ocr(pdf_file, output_file, poppler_path):
    # Create a directory for saving images
    if not os.path.exists('pdf_images'):
        os.makedirs('pdf_images')

    # Convert PDF pages to images, specifying the Poppler path
    images = convert_from_path(pdf_file, poppler_path=poppler_path)

    # Tesseract yolunu belirt
    pytesseract.pytesseract.tesseract_cmd = r'E:\Tesseract-OCR\tesseract.exe'
    # Iterate over images and apply OCR
    text_content = ""
    for i, image in enumerate(images):
        image_path = f"pdf_images/page_{i + 1}.png"
        image.save(image_path, 'PNG')

        # Perform OCR on the image
        text = pytesseract.image_to_string(Image.open(image_path))
        text_content += text

    # Save the extracted text to a file
    with open(output_file, 'w', encoding='utf-8') as text_file:
        text_file.write(text_content)

    print(f"OCR completed. Text saved to {output_file}")

# Usage example
pdf_to_text_with_ocr('a.pdf', 'output.txt', poppler_path=r'E:\poppler-24.08.0\Library\bin')
```
