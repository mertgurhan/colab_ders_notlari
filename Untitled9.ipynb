```python
import io
import json
import os
import re
import nltk
from PIL import Image
import fitz  # PyMuPDF
import pytesseract
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer

# Tesseract executable path
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# NLTK için gerekli verileri indir
nltk.download('stopwords')
nltk.download('omw-1.4')
nltk.download('wordnet')  # WordNet verisini indiriyoruz

# Stopwords ve lemmatizer tanımları
stop_words = set(stopwords.words('english'))
stemmer = SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    """
    OCR'den alınan metni temizlemek ve düzenlemek için ön işleme fonksiyonu.
    """
    # Satır sonlarındaki boşlukları kaldır
    text = text.strip()

    # Gereksiz boşlukları kaldır
    text = re.sub(r'\s+', ' ', text)

    # Belirli sembolleri temizleme (örneğin, yanlış tanınmış semboller)
    text = re.sub(r'[^\w\s.,!?%&]', '', text)

    # Büyük-küçük harf dönüşümü
    text = text.lower()

    # Durdurma kelimelerini kaldırma
    text = ' '.join(word for word in text.split() if word not in stop_words)

    # Kelimelerin köklerini bulma (stem)
    text = ' '.join(stemmer.stem(word) for word in text.split())

    # Lemmatization (kelime köklerine indirme) kullanmak yerine,
    # Eğer WordNet indirilemiyorsa sadece stemming kullanacak şekilde değiştirdik.
    try:
        text = ' '.join(lemmatizer.lemmatize(word) for word in text.split())
    except Exception as e:
        print(f"Warning: Lemmatization failed with error {e}. Using stemming instead.")
        # Eğer lemmatization başarısız olursa, stemming kullanmaya devam et
        text = ' '.join(stemmer.stem(word) for word in text.split())

    return text

def split_into_paragraphs(text):
    """
    Metni paragraflara böler. İki veya daha fazla yeni satır karakterine göre bölme yapılır.
    """
    paragraphs = re.split(r'\n\s*\n', text)
    return [para.strip() for para in paragraphs if para.strip()]

def pdf_to_text_with_ocr_pymupdf(pdf_file, output_file):
    try:
        with open(output_file, 'w', encoding='utf-8') as text_file:
            # PDF dosyasını PyMuPDF ile aç
            pdf_document = fitz.open(pdf_file)
            all_paragraphs = []  # Tüm paragrafları depolamak için bir liste
            for page_number in range(len(pdf_document)):
                # Sayfayı resme dönüştür
                page = pdf_document.load_page(page_number)
                pix = page.get_pixmap(dpi=200)  # Daha iyi OCR doğruluğu için DPI ayarı

                # Resmi byte dizisine kaydet
                img_buffer = io.BytesIO(pix.tobytes("png"))
                img_buffer.seek(0)

                # OCR ile resmi metne dönüştür
                try:
                    with Image.open(img_buffer) as img:
                        raw_text = pytesseract.image_to_string(img)

                        # Veri ön işleme
                        processed_text = preprocess_text(raw_text)

                        # Metni paragraflara böl
                        paragraphs = split_into_paragraphs(processed_text)
                        all_paragraphs.extend(paragraphs)  # Paragrafları kaydet

                        # Temizlenmiş paragrafları dosyaya yaz
                        for para in paragraphs:
                            text_file.write(para + "\n\n")
                except Exception as e:
                    print(f"Error processing page {page_number + 1}: {e}")

                print(f"Processed page {page_number + 1}")

            # Paragrafları JSON formatında da kaydetmek isterseniz:
            json_output_file = os.path.splitext(output_file)[0] + ".json"
            with open(json_output_file, 'w', encoding='utf-8') as json_file:
                json.dump(all_paragraphs, json_file, ensure_ascii=False, indent=4)

            print(f"OCR completed successfully. Text saved to '{output_file}'. JSON saved to '{json_output_file}'.")

    except FileNotFoundError:
        print(f"Error: The file '{pdf_file}' was not found. Please check the file path.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

def process_multiple_pdfs(pdf_files):
    for pdf_file in pdf_files:
        # PDF dosyasından temel ismi alarak çıkış dosyasını oluştur
        base_name = os.path.splitext(os.path.basename(pdf_file))[0]
        output_file = f"{base_name}_processed.txt"

        print(f"Processing '{pdf_file}' and saving output to '{output_file}'")
        pdf_to_text_with_ocr_pymupdf(pdf_file, output_file)

# Kullanım örneği
pdf_files = [
    r'/Users/mertgurhan/Desktop/Deep Neural Network Language Models.pdf'
    # Diğer PDF dosyalarınızı buraya ekleyebilirsiniz
]
process_multiple_pdfs(pdf_files)
```
