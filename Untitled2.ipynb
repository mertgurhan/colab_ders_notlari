```python
import tkinter as tk

def buton_tiklandi():
    label.config(text="Merhaba, " + isim_giris.get())


pencere = tk.Tk()
pencere.title("Arayüz")


label = tk.Label(pencere, text="Adınızı girin:")
label.pack(pady=10)


isim_giris = tk.Entry(pencere)
isim_giris.pack(pady=10)

buton = tk.Button(pencere, text="hellü", command=buton_tiklandi)
buton.pack(pady=10)

pencere.mainloop()
```

```python
import ipywidgets as widgets
from IPython.display import display

def buton_tiklandi(b):
    label.value = "Merhaba, " + isim_giris.value

label = widgets.Label(value="Adınızı girin:")
display(label)

isim_giris = widgets.Text()
display(isim_giris)

buton = widgets.Button(description="hellü")
buton.on_click(buton_tiklandi)
display(buton)
```

```python
import ipywidgets as widgets
from IPython.display import display

listbox = widgets.Select(
    options=[],
    value=None,
    description="List:",
    disabled=False
)

entry = widgets.Text(description="Add Item:")


button_add = widgets.Button(description="Add")
button_clear = widgets.Button(description="Clear")


def add_item(b):
    listbox.options += (entry.value,)
    entry.value = ""


def clear_list(b):
    listbox.options = []


button_add.on_click(add_item)
button_clear.on_click(clear_list)
display(listbox, entry, button_add, button_clear)
```
