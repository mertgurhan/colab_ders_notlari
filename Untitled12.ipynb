```python
import numpy as np
import matplotlib.pyplot as plt

# Kısıt doğrularını çizmek için X1 aralığını belirleyelim
X1 = np.linspace(0, 30, 100)

# Kısıtları belirleyelim
X2_1 = (100 - 4 * X1) / 2  # 4X1 + 2X2 ≤ 100  ==> X2 = (100 - 4X1) / 2
X2_2 = (200 - 10 * X1) / 8  # 10X1 + 8X2 ≤ 200 ==> X2 = (200 - 10X1) / 8

# Feasible bölgeyi belirlemek için minimum noktaları alalım
X2_feasible = np.minimum(X2_1, X2_2)
X2_feasible = np.maximum(X2_feasible, 0)  # Negatif değerleri sıfırla

# Grafik çizimi
plt.figure(figsize=(8, 6))
plt.plot(X1, X2_1, label=r'$4X_1 + 2X_2 \leq 100$', linestyle='dashed', color='blue')
plt.plot(X1, X2_2, label=r'$10X_1 + 8X_2 \leq 200$', linestyle='dashed', color='red')

# Feasible bölgeyi renklendirme
plt.fill_between(X1, X2_feasible, color='lightgray', alpha=0.5, label='Feasible Bölge')

# Eksenleri ve etiketleri ayarla
plt.xlim(0, 30)
plt.ylim(0, 30)
plt.xlabel(r'$X_1$ (Ürün 1)')
plt.ylabel(r'$X_2$ (Ürün 2)')
plt.title('Doğrusal Programlama - Grafik Yöntemi')
plt.legend()
plt.grid()

# Grafik gösterimi
plt.show()
```
