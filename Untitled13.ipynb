```python
# Soru: A = [6.4, 2.9, 4.3, 1.3] ile B = [5.1, 3.5, 1.4, 0.2] arasındaki Öklid mesafesini hesapla.
```

```python

A = [6.4, 2.9, 4.3, 1.3]
B = [5.1, 3.5, 1.4, 0.2]


def calculate_distance(ins1, ins2):
  sum=0
  for index in range(len(ins1)):
    sum+=(ins1[index]-ins2[index])**2
  sqr=sum**0.5
  return sqr
distance=calculate_distance(A,B)
print(distance)
```

```python
import numpy as np

A_np = np.array(A)
B_np = np.array(B)
dist_np = np.linalg.norm(A_np - B_np)
print("A2 Cevap (NumPy mesafe):", round(dist_np, 4))  # ≈ 3.4161
```
