```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import numpy as np

# Eğitim verisi (örnek)
data = [
    ("Merhaba", "Merhaba! Size nasıl yardımcı olabilirim?"),
    ("Nasılsın?", "Ben bir yapay zeka modeliyim, duygularım yok."),
    ("Python nedir?", "Python, yüksek seviyeli bir programlama dilidir."),
    ("Hoşça kal", "Hoşça kal! İyi günler dilerim.")
]

# Veriyi ayır
questions, answers = zip(*data)

# Vektörizasyon
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(questions)
y = np.array(answers)

# Modeli oluştur ve eğit
model = MultinomialNB()
model.fit(X, y)

def get_response(user_input):
    user_input_vectorized = vectorizer.transform([user_input])
    response = model.predict(user_input_vectorized)
    return response[0]

def main():
    print("Chatbot'a hoş geldiniz! 'çık' yazana kadar sohbet edebilirsiniz.")

    while True:
        user_input = input("Kullanıcı: ")
        if user_input.lower() == "çık":
            print("Chatbot: Hoşça kal!")
            break
        response = get_response(user_input)
        print(f"Chatbot: {response}")

if __name__ == "__main__":
    main()
```

```python
import pytesseract
from pdf2image import convert_from_path

def pdf_to_text(pdf_path):
    # PDF dosyasının sayfalarını görüntüye dönüştürme
    pages = convert_from_path(pdf_path)

    text = ""
    for page in pages:
        # Her sayfayı metne çevirme
        text += pytesseract.image_to_string(page)

    return text

if __name__ == "__main__":
    pdf_path = 'a.pdf'  # PDF dosyanızın yolu
    extracted_text = pdf_to_text(pdf_path)
    print(extracted_text)
```
